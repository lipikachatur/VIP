```python
from astropy.coordinates import SkyCoord
from astropy import units as u
from astropy.table import Table

# Convert RA and DEC to SkyCoord objects
main_coords = SkyCoord(ra=main_table['RA']*u.degree, dec=main_table['DEC']*u.degree)
mag_coords = SkyCoord(ra=magnitude_table['RA']*u.degree, dec=magnitude_table['DEC']*u.degree)

# Matching the catalogs
idx, d2d, _ = main_coords.match_to_catalog_sky(mag_coords)

# Filter matches with a max separation, e.g., 0.2 arcseconds
max_sep = 0.2 * u.arcsec
close_matches = d2d < max_sep

# Create a new table with only the close matches
matched_table = main_table[close_matches]
matched_table['Matched_Index'] = idx[close_matches]
matched_table['Separation'] = d2d[close_matches].to(u.arcsec)

# Optionally add matched data from magnitude_table
for col in magnitude_table.columns:
    matched_table[col + '_matched'] = magnitude_table[col][idx[close_matches]]

# Save or display your matched table
#matched_table.write('matched_table.fits', format='fits')
```
